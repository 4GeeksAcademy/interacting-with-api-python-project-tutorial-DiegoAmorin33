```python
client = MusicBrainzClient()
artists = client.search_artist("Coldplay")

len(artists)


```

```python
import pandas as pd

df = pd.DataFrame(artists)
df.head()
```

```python
import matplotlib.pyplot as plt

df["score"] = pd.to_numeric(df["score"], errors="coerce")

plt.figure(figsize=(8,5))
plt.scatter(range(len(df)), df["score"])
plt.xlabel("Resultados de búsqueda")
plt.ylabel("Score")
plt.title("Resultados MusicBrainz para 'Coldplay'")
plt.show()
```
